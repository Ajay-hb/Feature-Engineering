```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("/content/diet_recommendation_dataset_1000.csv")
```

```python
df
```

```python
x=df.drop(columns=["Diet"])#target data
y=df["Diet"]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=365,stratify=y)
```

```python
ss=StandardScaler()
```

```python
le=LabelEncoder()
```

```python
xtrain_scaled=ss.fit_transform(xtrain)
xtest_scaled=ss.transform(xtest)
```

```python
ytrain_scaled=le.fit_transform(ytrain)
ytest_scaled=le.transform(ytest)
```

```python
xtrain_scaled.shape
```

```python
xtest_scaled.shape
```

```python
ytest_scaled.shape
```

```python
ytrain_scaled.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
model=LogisticRegression(max_iter=1000)
```

```python
ffs=SequentialFeatureSelector(
    model,
    n_features_to_select=5,
    direction='forward',
    scoring="accuracy",
    cv=5
)    #forward feature selection
```

```python
ffs.fit(xtrain_scaled,ytrain_scaled)
```

```python
ffs.get_support()
```

```python
x.columns
```

```python
ffs_features=x.columns[ffs.get_support()]
```

```python
ffs_features
```

```python
bfs=SequentialFeatureSelector(
    model,
    n_features_to_select=5,
    direction='backward',
    scoring="accuracy",
    cv=5
)    #backward feature selection
```

```python
bfs.fit(xtrain_scaled,ytrain_scaled)
```

```python
bfs_features=x.columns[bfs.get_support()]
```

```python
bfs_features
```

```python
ffs_features
```

```python
xtrain_bfs = xtrain[bfs_features]
xtest_bfs = xtest[bfs_features]

# Scale the BFS selected features
ss_bfs = StandardScaler()
xtrain_scaled_bfs = ss_bfs.fit_transform(xtrain_bfs)
xtest_scaled_bfs = ss_bfs.transform(xtest_bfs)

# Train a Logistic Regression model with BFS selected features
lr_bfs = LogisticRegression(max_iter=1000)
lr_bfs.fit(xtrain_scaled_bfs, ytrain_scaled)

# Calculate accuracy
accuracy_bfs = lr_bfs.score(xtest_scaled_bfs, ytest_scaled)
print(f"Accuracy with Backward Feature Selection: {accuracy_bfs:.4f}")
```

```python
xtrain_ffs = xtrain[ffs_features]
xtest_ffs = xtest[ffs_features]

# Scale the FFS selected features
ss_ffs = StandardScaler()
xtrain_scaled_ffs = ss_ffs.fit_transform(xtrain_ffs)
xtest_scaled_ffs = ss_ffs.transform(xtest_ffs)

# Train a Logistic Regression model with FFS selected features
lr_ffs = LogisticRegression(max_iter=1000)
lr_ffs.fit(xtrain_scaled_ffs, ytrain_scaled)

# Calculate accuracy
accuracy_ffs = lr_ffs.score(xtest_scaled_ffs, ytest_scaled)
print(f"Accuracy with Forward Feature Selection: {accuracy_ffs:.4f}")
```

```python
xtrain_final = xtrain.drop(columns=['Height_cm'])
xtest_final = xtest.drop(columns=['Height_cm'])

# Scale data with Weight_kg (Height_cm excluded)
ss_with_weight = StandardScaler()
xtrain_scaled_with_weight = ss_with_weight.fit_transform(xtrain_final)
xtest_scaled_with_weight = ss_with_weight.transform(xtest_final)

# Train KNN model with Weight_kg
knn_with_weight = KNeighborsClassifier(n_neighbors=5)
knn_with_weight.fit(xtrain_scaled_with_weight, ytrain_scaled)
accuracy_with_weight = knn_with_weight.score(xtest_scaled_with_weight, ytest_scaled)
print(f"KNN Model Accuracy (with Weight_kg, without Height_cm): {accuracy_with_weight:.4f}")

# Prepare data without Weight_kg (and without Height_cm)
xtrain_no_weight_no_height = xtrain_final.drop(columns=['Weight_kg'])
xtest_no_weight_no_height = xtest_final.drop(columns=['Weight_kg'])

# Scale data without Weight_kg (Height_cm excluded)
ss_no_weight = StandardScaler()
xtrain_scaled_no_weight_no_height = ss_no_weight.fit_transform(xtrain_no_weight_no_height)
xtest_scaled_no_weight_no_height = ss_no_weight.transform(xtest_no_weight_no_height)

# Train KNN model without Weight_kg
knn_no_weight = KNeighborsClassifier(n_neighbors=5)
knn_no_weight.fit(xtrain_scaled_no_weight_no_height, ytrain_scaled)
accuracy_no_weight = knn_no_weight.score(xtest_scaled_no_weight_no_height, ytest_scaled)
print(f"KNN Model Accuracy (without Weight_kg and Height_cm): {accuracy_no_weight:.4f}")
```
